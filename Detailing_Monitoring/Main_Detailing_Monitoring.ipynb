```python
def show_structure(obj, indent=0, max_depth=3):
    pad = "  " * indent
    if indent > max_depth:
        return
    if isinstance(obj, dict):
        for k, v in obj.items():
            print(f"{pad}{k}: {type(v).__name__}", end="")
            if isinstance(v, (list, dict)):
                print(f" (len={len(v)})")
                show_structure(v, indent + 1, max_depth)
            else:
                print(f" = {repr(v)[:60]}")
    elif isinstance(obj, list) and obj:
        print(f"{pad}[0] -> {type(obj[0]).__name__}")
        show_structure(obj[0], indent + 1, max_depth)

show_structure(data)
```

```python
import re

DETAIL = ["activitydetail", "speed", "rpm", "eta", "remainingdistance", "cargoremarks"]

# ---------- 1. pivot aktivitas per project (trip sendiri) ----------
ev = df_ev.dropna(subset=["activity_name"]).copy()
ev["act"] = ev["activity_port"] + " | " + ev["activity_name"]
ev = ev.sort_values(["project", "started_at", "id"])

g = ev.groupby(["project", "act"])
res = (g.agg(Start=("started_at", "first"),
             Last=("started_at", "last"),
             Count=("id", "size"))
         .join(g[DETAIL].last())
         .reset_index())

order = (ev.drop_duplicates("act")
           .sort_values("activities")["act"].tolist())

parts = []
for act in order:
    s = res[res["act"] == act].set_index("project")
    b = pd.DataFrame(index=s.index)
    b[f"{act} | Start"] = s["Start"]
    if s["Count"].max() > 1:
        b[f"{act} | Last"] = s["Last"]
        b[f"{act} | Count"] = s["Count"]
    for c in DETAIL:
        if s[c].notna().any():
            b[f"{act} | {c}"] = s[c]
    parts.append(b)

wide = pd.concat(parts, axis=1)

# ---------- 2. info project ----------
proj_cols = [
    "id", "code", "remarks", "owner", "started_at", "expected_at",
    "closed_at", "closed_by", "tugboatel.name", "bargeel.name",
    "bargeel.feet", "tugboatel.feet", "customerel.shortname",
    "customerel.fullname", "cargoel.name", "polel.name", "polel.shortname",
    "podel.name", "podel.shortname", "podpreviousel.name", "spal", "si",
    "sidate", "price", "prorata", "bbm", "demmurage", "ondemmurage",
    "sapb1", "projectremarks", "polagen", "podagen",
    "polstartdemmurage", "polenddemmurage",
    "podstartdemmurage", "podenddemmurage",
    "downtimeamount",
]
proj_cols = [c for c in proj_cols if c in df_proj.columns]

info = (df_proj[proj_cols]
        .rename(columns={"id": "project_id", "code": "Project Code"})
        .copy())

# ---------- 3. cari trip sebelumnya dari Project Code ----------
pat = info["Project Code"].str.extract(r"^(.*)-(\d{4})-(\d+)$")
info["_prefix"] = pat[0]
info["_year"] = pat[1].astype(float)
info["_trip"] = pat[2].astype(float)

info = info.sort_values(["_prefix", "_year", "_trip"]).reset_index(drop=True)
info["T1 Project Code"] = info.groupby("_prefix")["Project Code"].shift(1)
info["_prev_id"] = info.groupby("_prefix")["project_id"].shift(1)

# ---------- 4. T1: hanya FAW Sailing To Finish & Actual Arrive Finish ----------
T1_SRC = {
    "T1 FAW Sailing To Finish": "POD | FAW Sailing To Finish | Start",
    "T1 Actual Arrive Finish":  "Finish | Actual Arrive Finish | Start",
}
for new, src in T1_SRC.items():
    info[new] = info["_prev_id"].map(wide[src]) if src in wide.columns else pd.NaT

# ---------- 5. T2: semua aktivitas trip sendiri ----------
wide_t2 = wide.add_prefix("T2 | ")

df_turn = (info.drop(columns=["_prefix", "_year", "_trip", "_prev_id"])
               .merge(wide_t2, left_on="project_id", right_index=True, how="left")
               .reset_index(drop=True))

print("Shape:", df_turn.shape)
print("Project Code unik?:", df_turn["Project Code"].is_unique)
print("Project tanpa trip sebelumnya:", df_turn["T1 Project Code"].isna().sum())
print("Kode yang tidak cocok pola:", info["_prefix"].isna().sum() if "_prefix" in info else pat[0].isna().sum())
df_turn[["Project Code", "T1 Project Code",
         "T1 FAW Sailing To Finish", "T1 Actual Arrive Finish"]].head(10)
```

```python
def col(name):
    return name if name in df_turn.columns else None

def days(a, b):
    """selisih b - a dalam hari (float); NaN kalau kolom/nilai tidak ada"""
    if col(a) is None or col(b) is None:
        return pd.Series(float("nan"), index=df_turn.index)
    return (df_turn[b] - df_turn[a]).dt.total_seconds() / 86400

POL_ARR   = "T2 | POL | Actual Arrive POL | Start"
POL_LS    = "T2 | POL | Loading Start | Start"
POL_LC    = "T2 | POL | Loading Complete | Start"
POL_FAW   = "T2 | POL | FAW Sailing To POD | Start"
POD_ARR   = "T2 | POD | Actual Arrive POD | Start"
POD_DS    = "T2 | POD | Discharge Start | Start"
POD_DC    = "T2 | POD | Discharge Complete | Start"
POD_FAW   = "T2 | POD | FAW Sailing To Finish | Start"
FIN_ARR   = "T2 | Finish | Actual Arrive Finish | Start"
T1_FAW    = "T1 FAW Sailing To Finish"
T1_ARR    = "T1 Actual Arrive Finish"

# jeda antar trip dan voyage berdasarkan FAW trip sebelumnya
df_turn["Voyage V.2 (hari)"]          = days(T1_FAW, POD_FAW)
df_turn["Jeda Finish T1 ke Arrive POL (hari)"] = days(T1_ARR, POL_ARR)

# fase di POL
df_turn["POL Loading (hari)"]         = days(POL_LS, POL_LC)
df_turn["POL Total Port (hari)"]      = days(POL_ARR, POL_FAW)

# pelayaran
df_turn["Sailing POL ke POD (hari)"]  = days(POL_FAW, POD_ARR)

# fase di POD
df_turn["POD Discharge (hari)"]       = days(POD_DS, POD_DC)
df_turn["POD Total Port (hari)"]      = days(POD_ARR, POD_FAW)

# pelayaran balik ke finish
df_turn["Sailing POD ke Finish (hari)"] = days(POD_FAW, FIN_ARR)

# total voyage = (Arrive POD - FAW POL) + (Arrive Finish - FAW POD)
df_turn["Total Voyage Days"] = (df_turn["Sailing POL ke POD (hari)"]
                                + df_turn["Sailing POD ke Finish (hari)"])

dur_cols = [c for c in df_turn.columns if c.endswith("(hari)") or c == "Total Voyage Days"]
print(df_turn[dur_cols].describe().T.round(2))

# cek kolom negatif (indikasi urutan waktu salah / salah input)
neg = (df_turn[dur_cols] < 0).sum()
print("\nJumlah nilai negatif per kolom:")
print(neg[neg > 0])

# cek 41 project tanpa T1
print("\nProject tanpa T1:")
no_t1 = df_turn[df_turn["T1 Project Code"].isna()]
print(no_t1[["Project Code", "tugboatel.name", "bargeel.name", "started_at"]]
      .to_string())
```

```python
import re
import numpy as np

# ---------- 1. nama kolom aman untuk SQL ----------
def snake(name):
    s = name.replace("T2 | ", "t2_").replace("T1 ", "t1_")
    s = s.replace("|", " ").replace("/", " ").replace(".", "_")
    s = re.sub(r"[^0-9a-zA-Z_ ]+", " ", s)
    s = re.sub(r"\s+", "_", s.strip()).lower()
    s = re.sub(r"_+", "_", s)
    return s

df_sb = df_turn.copy()
colmap = {c: snake(c) for c in df_sb.columns}

# pastikan tidak ada nama kembar
dup = pd.Series(list(colmap.values())).duplicated(keep=False)
assert not dup.any(), pd.Series(list(colmap.values()))[dup].tolist()

df_sb = df_sb.rename(columns=colmap)

# ---------- 2. tipe data ----------
dt_cols  = [c for c in df_sb.columns if pd.api.types.is_datetime64_any_dtype(df_sb[c])]
for c in ["started_at", "expected_at", "closed_at", "sidate", "polstartdemmurage"]:
    pass  # kolom tanggal dari projectel masih string, di-parse di bawah

for c in ["started_at", "expected_at", "closed_at", "sidate"]:
    if c in df_sb.columns:
        df_sb[c] = pd.to_datetime(df_sb[c], errors="coerce")
        if c not in dt_cols:
            dt_cols.append(c)

num_cols = [c for c in ["price", "demmurage", "bbm", "bargeel_feet", "tugboatel_feet"]
            if c in df_sb.columns]
for c in num_cols:
    df_sb[c] = pd.to_numeric(df_sb[c], errors="coerce")

# kolom detail hasil pivot yang isinya angka (speed, rpm, remainingdistance)
for c in df_sb.columns:
    if c.endswith(("_speed", "_rpm", "_remainingdistance", "_count")):
        df_sb[c] = pd.to_numeric(df_sb[c], errors="coerce")

# kolom eta berupa waktu
for c in [c for c in df_sb.columns if c.endswith("_eta")]:
    df_sb[c] = pd.to_datetime(df_sb[c], errors="coerce")
    if c not in dt_cols:
        dt_cols.append(c)

# ---------- 3. NaT/NaN -> None, datetime -> string ISO ----------
def to_json_safe(df):
    out = df.copy()
    for c in out.columns:
        if pd.api.types.is_datetime64_any_dtype(out[c]):
            out[c] = out[c].dt.strftime("%Y-%m-%dT%H:%M:%S")
    out = out.astype(object).where(out.notna(), None)
    return out

payload = to_json_safe(df_sb).to_dict(orient="records")
json.dumps(payload[:50])  # tes: pastikan bisa diserialisasi

print("Kolom   :", df_sb.shape[1])
print("Baris   :", len(payload))
print("Kunci upsert unik:", df_sb["project_code"].is_unique)
print("\nContoh 1 baris (yang tidak null):")
print({k: v for k, v in payload[0].items() if v is not None})
```

```python
import os, time
from datetime import datetime, timezone

SUPABASE_URL = os.environ["SUPABASE_URL"].rstrip("/")
SUPABASE_KEY = os.environ["SUPABASE_KEY"]   # service_role key
TABLE        = "monitoring_turnaround_v2"
KEY          = "project_code"
BATCH        = 200
MIN_RATIO    = 0.90     # guard: jumlah baru minimal 90% dari jumlah di tabel
DRY_RUN      = True     # ganti False kalau sudah yakin

BASE = f"{SUPABASE_URL}/rest/v1/{TABLE}"
HDR  = {
    "apikey": SUPABASE_KEY,
    "Authorization": f"Bearer {SUPABASE_KEY}",
    "Content-Type": "application/json",
}

def req(method, url, retries=3, **kw):
    for i in range(retries):
        try:
            r = requests.request(method, url, headers={**HDR, **kw.pop("headers", {})},
                                 timeout=120, **kw)
            if r.status_code < 300:
                return r
            if r.status_code < 500 and r.status_code != 429:
                raise RuntimeError(f"{r.status_code}: {r.text[:500]}")
            err = f"{r.status_code}: {r.text[:200]}"
        except requests.RequestException as e:
            err = str(e)
        time.sleep(2 ** i)
    raise RuntimeError(f"Gagal setelah {retries}x: {err}")

def existing_keys():
    keys, offset, page = [], 0, 1000
    while True:
        r = req("GET", BASE, params={"select": KEY, "order": KEY,
                                     "limit": page, "offset": offset})
        rows = r.json()
        keys += [x[KEY] for x in rows]
        if len(rows) < page:
            break
        offset += page
    return keys

# ---------- 1. siapkan payload ----------
now = datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")
rows = [{**p, "synced_at": now} for p in payload]

new_keys = [r[KEY] for r in rows]
assert all(new_keys), "Ada project_code kosong"
assert len(new_keys) == len(set(new_keys)), "project_code duplikat"

# ---------- 2. guard ----------
old_keys = existing_keys()
print(f"Baris di tabel: {len(old_keys)} | baris baru: {len(rows)}")

if not rows:
    raise SystemExit("ABORT: payload kosong")
if old_keys and len(rows) < MIN_RATIO * len(old_keys):
    raise SystemExit(f"ABORT: jumlah baru turun drastis "
                     f"({len(rows)} < {MIN_RATIO:.0%} x {len(old_keys)})")

orphans = sorted(set(old_keys) - set(new_keys))
print("Akan di-upsert :", len(rows))
print("Akan dihapus   :", len(orphans), orphans[:10])

if DRY_RUN:
    print("\nDRY_RUN=True -> tidak ada yang ditulis.")
else:
    # ---------- 3. upsert per batch ----------
    for i in range(0, len(rows), BATCH):
        chunk = rows[i:i + BATCH]
        req("POST", BASE,
            params={"on_conflict": KEY},
            headers={"Prefer": "resolution=merge-duplicates,return=minimal"},
            data=json.dumps(chunk))
        print(f"Upsert {min(i + BATCH, len(rows))}/{len(rows)}")

    # ---------- 4. hapus yatim (hanya setelah upsert sukses) ----------
    for i in range(0, len(orphans), 100):
        part = orphans[i:i + 100]
        in_list = ",".join('"' + c.replace('"', '') + '"' for c in part)
        req("DELETE", BASE, params={KEY: f"in.({in_list})"},
            headers={"Prefer": "return=minimal"})
    print(f"Dihapus {len(orphans)} baris yatim")

    # ---------- 5. verifikasi ----------
    final = existing_keys()
    print(f"\nSelesai. Baris di tabel: {len(final)} (target {len(rows)})")
    assert set(final) == set(new_keys), "Isi tabel tidak sama dengan payload"
```
